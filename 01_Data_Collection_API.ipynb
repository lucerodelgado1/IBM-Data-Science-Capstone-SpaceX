# 1. Data Collection - SpaceX API

This notebook collects historical SpaceX launch data from the SpaceX REST API and converts the JSON response into a pandas DataFrame.

```python
import requests
import pandas as pd
import json
```

## Get data from the SpaceX API

```python
url = 'https://api.spacexdata.com/v4/launches/past'
response = requests.get(url)
print('Status code:', response.status_code)
```

```python
data = response.json()
print('Number of launch records:', len(data))
```

## Convert JSON response to a DataFrame

```python
df = pd.json_normalize(data)
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
df.head()
```

## Inspect the available variables

```python
df.columns.tolist()
```

## Save the collected data

```python
df.to_csv('spacex_launches_raw.csv', index=False)
print('File saved: spacex_launches_raw.csv')
```
