# Data Collection – SpaceX API

Collect Falcon 9 launch data from the SpaceX REST API and convert JSON into pandas. A public course-compatible dataset is used if the live endpoint is temporarily unavailable.

```python
import requests
import pandas as pd

API_URL = "https://api.spacexdata.com/v4/launches/past"
response = requests.get(API_URL, timeout=30)
print("Status code:", response.status_code)
print("Content type:", response.headers.get("content-type"))

if response.ok and "json" in response.headers.get("content-type","").lower():
    data = response.json()
    df_api = pd.json_normalize(data)
    print("API records:", len(df_api))
    display(df_api.head())
else:
    print("The live API did not return JSON; loading the public Falcon 9 dataset.")
```

```python
DATA_URL = "https://raw.githubusercontent.com/asifikbalpro/Applied-Data-Science-Capstone/main/dataset_part_1.csv"
df = pd.read_csv(DATA_URL)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
display(df.head())
```
