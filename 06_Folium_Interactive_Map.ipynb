# Interactive Visual Analytics – Folium

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster

df = pd.read_csv("https://raw.githubusercontent.com/asifikbalpro/Applied-Data-Science-Capstone/main/dataset_part_2.csv")
site_coords = df.groupby(["LaunchSite","Latitude","Longitude"], as_index=False).size()

m = folium.Map(location=[29.0, -95.0], zoom_start=4)
cluster = MarkerCluster().add_to(m)

for _, r in site_coords.iterrows():
    folium.Marker(
        [r["Latitude"], r["Longitude"]],
        popup=f'{r["LaunchSite"]}<br>Launches: {int(r["size"])}'
    ).add_to(cluster)

m
```

```python
outcome_map = folium.Map(location=[29.0, -95.0], zoom_start=4)
for _, r in df.dropna(subset=["Latitude","Longitude"]).iterrows():
    color = "green" if r["Class"] == 1 else "red"
    folium.CircleMarker(
        location=[r["Latitude"], r["Longitude"]],
        radius=5, color=color, fill=True, fill_opacity=.7,
        popup=f'Flight {r["FlightNumber"]} | Class {r["Class"]}'
    ).add_to(outcome_map)
outcome_map
```
