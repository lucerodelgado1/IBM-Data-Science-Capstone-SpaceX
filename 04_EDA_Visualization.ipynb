# Exploratory Data Analysis – Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("https://raw.githubusercontent.com/asifikbalpro/Applied-Data-Science-Capstone/main/dataset_part_2.csv")
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
```

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x="LaunchSite")
plt.title("Falcon 9 launches by launch site")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df, x="FlightNumber", y="PayloadMass", hue="Class")
plt.title("Flight number vs payload mass by landing outcome")
plt.tight_layout()
plt.show()
```

```python
success_by_orbit = df.groupby("Orbit")["Class"].mean().sort_values(ascending=False)
plt.figure(figsize=(9,5))
success_by_orbit.plot(kind="bar")
plt.ylabel("Landing success rate")
plt.title("Landing success rate by orbit")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

```python
annual = df.groupby("Year")["Class"].mean()
plt.figure(figsize=(10,5))
annual.plot(marker="o")
plt.ylabel("Landing success rate")
plt.title("Annual landing success trend")
plt.grid(alpha=.2)
plt.tight_layout()
plt.show()
```
