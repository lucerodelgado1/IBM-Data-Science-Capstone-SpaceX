# Exploratory Data Analysis – SQL

SQLite reproduces the required SQL analysis.

```python
import pandas as pd
import sqlite3

df = pd.read_csv("https://raw.githubusercontent.com/asifikbalpro/Applied-Data-Science-Capstone/main/dataset_part_2.csv")
conn = sqlite3.connect("spacex.db")
df.to_sql("SPACEXTBL", conn, if_exists="replace", index=False)
```

```python
queries = {
"Unique launch sites": "SELECT DISTINCT LaunchSite FROM SPACEXTBL",
"Minimum payload": "SELECT MIN(PayloadMass) AS Min_Payload FROM SPACEXTBL",
"Total payload": "SELECT SUM(PayloadMass) AS Total_Payload_Mass FROM SPACEXTBL",
"Launch count by site": "SELECT LaunchSite, COUNT(*) AS Launches FROM SPACEXTBL GROUP BY LaunchSite",
"Success by site": "SELECT LaunchSite, AVG(Class) AS Success_Rate FROM SPACEXTBL GROUP BY LaunchSite"
}
for name, q in queries.items():
    print("\n---", name, "---")
    display(pd.read_sql_query(q, conn))
```
