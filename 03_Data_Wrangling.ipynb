# Data Wrangling – SpaceX Falcon 9

```python
import pandas as pd
import numpy as np

df = pd.read_csv("https://raw.githubusercontent.com/asifikbalpro/Applied-Data-Science-Capstone/main/dataset_part_1.csv")
df["Class"] = df["Outcome"].astype(str).str.startswith("True").astype(int)
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["PayloadMass"] = pd.to_numeric(df["PayloadMass"], errors="coerce")
df["PayloadMass"] = df["PayloadMass"].fillna(df["PayloadMass"].median())

features = ["FlightNumber","Date","PayloadMass","Orbit","LaunchSite","Flights",
            "GridFins","Reused","Legs","Block","ReusedCount","Serial",
            "Longitude","Latitude","Class"]
clean_df = df[features].copy()

print("Clean shape:", clean_df.shape)
print("Missing values:", clean_df.isnull().sum().sum())
display(clean_df.head())
```

```python
model_df = pd.get_dummies(clean_df, columns=["Orbit","LaunchSite","Serial"], drop_first=True)
print("Model-ready shape:", model_df.shape)
display(model_df.head())
clean_df.to_csv("spacex_clean.csv", index=False)
model_df.to_csv("spacex_model_ready.csv", index=False)
```
